https://leetcode.com/problems/group-anagrams

```python
strs = ["eat", "tea", "tan", "ate", "nat", "bat"]

res = []

group = []

sorted_strs = [sorted(string) for string in strs]

print(sorted_strs)

for string in strs:

    if sorted(string) in sorted_strs:
        group.append(string)
    else:
        res.append(group)
        group = []

res
```
