https://leetcode.com/problems/group-anagrams

```python
class Solution:
    def groupAnagrams(self, strs: list[str]) -> list[list[str]]:
        res = {}

        for string in strs:
            key = "".join(sorted(string))
            try:
                res[key].append(string)
            except KeyError:
                res[key] = [string]

        return list(res.values())
```
